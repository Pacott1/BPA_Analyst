## Ejecución del proceso automatizado

La función `frankfurter_wrapper()` ejecuta el flujo completo de automatización del tipo de cambio EUR/USD.

Al ejecutarla, el programa:

1. Consulta automáticamente la API de Frankfurter.
2. Obtiene la fecha y el tipo de cambio EUR/USD disponible.
3. Comprueba si existe un histórico previo en `data/tipo_cambio_EUR_USD.csv`.
4. Si no existe, crea el archivo con el primer registro.
5. Si existe, añade el nuevo dato y calcula la variación porcentual respecto al último registro.
6. Evita añadir registros duplicados si la fecha ya se encuentra almacenada.
7. Muestra por consola el resultado de la ejecución.
